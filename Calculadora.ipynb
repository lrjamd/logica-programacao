#Calculadora Inteligente

```python
def calculadora(): # definição da função
    while True: # laço (true faz com que a função seja infinita até o break)
        # Solicita a operação do usuário
        operacao = input("Digite a operação (+, -, *, /) ou 'sair' para finalizar: ").strip() #strip remove espaços de digitação

        # Encerra o programa se o usuário digitar 'sair'
        if operacao == 'sair':
            print("Encerrando a calculadora...")
            break

        # Verifica se a operação é válida
        if operacao in ['+', '-', '*', '/']:
            try: # bloco try-except para tratar erros
                # Solicita os números para a operação
                num1 = float(input("Digite o primeiro número: "))
                num2 = float(input("Digite o segundo número: "))

                # Realiza a operação de acordo com a operação escolhida
                if operacao == '+':
                    resultado = num1 + num2
                elif operacao == '-':
                    resultado = num1 - num2
                elif operacao == '*':
                    resultado = num1 * num2
                elif operacao == '/':
                    if num2 != 0:
                        resultado = num1 / num2
                    else:
                        resultado = "Erro! Divisão por zero."

                # Exibe o resultado
                print(f"Resultado: {resultado}")
            except ValueError:
                print("Erro! Por favor, insira números válidos.")
        else:
            print("Operação inválida! Tente novamente.")

# Chama a função da calculadora
calculadora()
```
